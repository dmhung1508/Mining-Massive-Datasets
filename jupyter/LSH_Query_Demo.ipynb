# Query Demo - Input Text to Similar Tweets


```python
from __future__ import annotations

import sys
from pathlib import Path

import pandas as pd
from IPython.display import display


def resolve_repo_root() -> Path:
    cwd = Path.cwd().resolve()
    candidates = [cwd, cwd.parent]
    for candidate in candidates:
        if (candidate / 'src').exists() and (candidate / 'jupyter' / 'output' / 'lsh').exists():
            return candidate
    raise FileNotFoundError('Could not find repo root from the current notebook location.')


REPO_ROOT = resolve_repo_root()
SRC_PATH = REPO_ROOT / 'src'
if str(SRC_PATH) not in sys.path:
    sys.path.insert(0, str(SRC_PATH))

from uk_russia_lsh.search import prepare_search_index, search_similar_tweets

ARTIFACT_DIR = REPO_ROOT / 'jupyter' / 'output' / 'lsh'
pd.set_option('display.max_colwidth', 140)
ARTIFACT_DIR
```

```python
QUERY_TEXT = 'Go fuck yourself, Putin! (English)'
TOP_K = 5
MIN_JACCARD = 0.5

prepare_search_index(artifact_dir=ARTIFACT_DIR)
results, metadata = search_similar_tweets(
    query_text=QUERY_TEXT,
    artifact_dir=ARTIFACT_DIR,
    top_k=TOP_K,
    min_jaccard=MIN_JACCARD,
)

display(pd.DataFrame([metadata]))
display(results)
```
